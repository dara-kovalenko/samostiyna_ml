```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             classification_report, ConfusionMatrixDisplay)

VARIANT = 'bean'
pd.set_option('display.width', 160); pd.set_option('display.max_columns', 20)

df = pd.read_csv('drybean.csv')
target = 'Class'
features = [c for c in df.columns if c not in ('Class', 'split')]

train, val, test = (df[df['split'] == s] for s in ('train', 'val', 'test'))
X_tr, y_tr = train[features], train[target]
X_va, y_va = val[features], val[target]
X_te, y_te = test[features], test[target]

#завдання 1
display(df.head()) 
display(df.describe())

class_counts = y_tr.value_counts()
print(class_counts)
class_counts.plot(kind='bar', color='cyan', edgecolor='black', figsize=(8, 5))
plt.title('Розподіл класів')
plt.xlabel('Класи')
plt.ylabel('К-сть прикладів')
plt.xticks(rotation=0) 
plt.show()

most_frequent_class = y_tr.mode()[0]
lazy_preds = [most_frequent_class] * len(y_va)
lazy_accuracy = accuracy_score(y_va, lazy_preds)
print("Точність:", round(lazy_accuracy, 3))
print("-------")

#завдання 2
tree_model = DecisionTreeClassifier(max_depth=3, random_state=0)
tree_model.fit(X_tr, y_tr)

acc_tr = tree_model.score(X_tr, y_tr)
acc_va = tree_model.score(X_va, y_va)
print("Точність навч вибірки", round(acc_tr, 3))
print("Точність валід вибірки", round(acc_va, 3))

plt.figure(figsize=(24, 14))
plot_tree(tree_model, feature_names=list(X_tr.columns), class_names=list(tree_model.classes_), filled=True)
plt.show()

#завдання 3
forest_model = RandomForestClassifier(n_estimators=100, random_state=0)
forest_model.fit(X_tr, y_tr)

models = {"Дерево рішень": tree_model, "Випадковий ліс": forest_model}
for name, model in models.items():
    preds = model.predict(X_va)
    acc = accuracy_score(y_va, preds)
    prec = precision_score(y_va, preds, average='macro', zero_division=0)
    rec = recall_score(y_va, preds, average='macro')
    print(name)
    print("Точність:", round(acc, 3))
    print("Влучність:", round(prec, 3))
    print("Повнота:", round(rec, 3))

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
ConfusionMatrixDisplay.from_estimator(tree_model, X_va, y_va, ax=axes[0], xticks_rotation=45)
axes[0].set_title('Матриця помилок дерева рішень')
ConfusionMatrixDisplay.from_estimator(forest_model, X_va, y_va, ax=axes[1], xticks_rotation=45)
axes[1].set_title('Матриця помилок випадкового лісу')
plt.tight_layout()
plt.show()

#завдання 4
depths = range(1, 21)
tree_train_acc, tree_val_acc = [], []
for d in depths:
    dt = DecisionTreeClassifier(max_depth=d, random_state=0)
    dt.fit(X_tr, y_tr)
    tree_train_acc.append(dt.score(X_tr, y_tr))
    tree_val_acc.append(dt.score(X_va, y_va))
plt.figure(figsize=(10, 5))
plt.plot(depths, tree_train_acc, label='Точність тренування', marker='o')
plt.plot(depths, tree_val_acc, label='Точність валідації', marker='s')
plt.title('Дерево рішень: точність від max_depth')
plt.xlabel('Глибина дерева')
plt.ylabel('Точність')
plt.xticks(depths)
plt.legend()
plt.grid(True)
plt.show()

n_estimators = [1, 5, 10, 25, 50, 100, 200, 300]
forest_val_acc = []

for n in n_estimators:
    rf = RandomForestClassifier(n_estimators=n, random_state=0, n_jobs=-1)
    rf.fit(X_tr, y_tr)
    forest_val_acc.append(rf.score(X_va, y_va))
plt.figure(figsize=(10, 5))
plt.plot(n_estimators, forest_val_acc, marker='o', color='green')
plt.title('Випадковий ліс: точність від n_estimators')
plt.xlabel('К-сть дерев')
plt.ylabel('Точність на валідації')
plt.grid(True)
plt.show()

best_depth_idx = np.argmax(tree_val_acc)
best_depth = depths[best_depth_idx]
print("Найкраща глибина дерева:", best_depth, "Точність валідації:", round(tree_val_acc[best_depth_idx], 3))

best_tree = DecisionTreeClassifier(max_depth=best_depth, random_state=0).fit(X_tr, y_tr)
best_forest = RandomForestClassifier(n_estimators=100, random_state=0, n_jobs=-1).fit(X_tr, y_tr)
print("Точність найкращого дерева на test:", round(best_tree.score(X_te,y_te),3))
print("Точність найкращого лісу(100 дерев) на test:", round(best_tree.score(X_te,y_te),3))

#завдання 5
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
tree_imp = pd.Series(best_tree.feature_importances_, index=X_tr.columns).sort_values()
tree_imp.plot(kind='barh', ax=axes[0], title='Важливість ознак по дереву рішень')
forest_imp = pd.Series(best_forest.feature_importances_, index=X_tr.columns).sort_values()
forest_imp.plot(kind='barh', ax=axes[1], title='Важливість ознак по випадковому лісу', color='green')
plt.tight_layout()
plt.show()

top_3_features = forest_imp.sort_values(ascending=False).head(3).index.tolist()
print("Топ-3 ознаки лісу:", top_3_features)

features_to_use = top_3_features
X_tr_mod = X_tr[features_to_use]
X_va_mod = X_va[features_to_use]
X_te_mod = X_te[features_to_use]
mod_forest = RandomForestClassifier(n_estimators=100, random_state=0, n_jobs=-1).fit(X_tr_mod, y_tr)
print("Повна модель (val):", round(best_forest.score(X_va, y_va), 3))
print("Повна модель (test):", round(best_forest.score(X_te, y_te), 3))
print("Змінена модель (val):", round(mod_forest.score(X_va_mod, y_va), 3))
print("Змінена модель (test):", round(mod_forest.score(X_te_mod, y_te), 3))
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
ConfusionMatrixDisplay.from_estimator(best_forest, X_va, y_va, ax=axes[0], xticks_rotation=45)
axes[0].set_title('Матриця повного лісу')
ConfusionMatrixDisplay.from_estimator(mod_forest, X_va_mod, y_va, ax=axes[1], xticks_rotation=45)
axes[1].set_title('Матриця зміненого лісу')
plt.show()

#завдання 6
median_sira = X_tr[y_tr == 'SIRA'].median()
median_dermason = X_tr[y_tr == 'DERMASON'].median()
mixed_example = (median_sira + median_dermason) / 2
my_examples = pd.DataFrame([median_sira, median_dermason, mixed_example], columns=X_tr.columns)

tree_preds = best_tree.predict(my_examples)
forest_preds = best_forest.predict(my_examples)
forest_probs = best_forest.predict_proba(my_examples).max(axis=1)
results = pd.DataFrame({'Приклад': ['Медіана SIRA', 'Медіана DERMASON', 'Суміш SIRA+DERMASON'], 'Прогноз дерева': tree_preds, 'Прогноз лісу': forest_preds, 'Ймовірність лісу': forest_probs})
display(results)

#завдання 7
n_root = tree_model.tree_.n_node_samples[0]
G_root = tree_model.tree_.impurity[0]
n_L1 = tree_model.tree_.n_node_samples[1]
G_L1 = tree_model.tree_.impurity[1]
n_R1 = tree_model.tree_.n_node_samples[2]
G_R1 = tree_model.tree_.impurity[2]

G_prime_root = (n_L1 / n_root) * G_L1 + (n_R1 / n_root) * G_R1
delta_G_root = G_root - G_prime_root

print("--- КОРІНЬ ---")
print(f"Індекс Джині кореня (G): {G_root:.4f}")
print(f"Зважений Джині (G'): {G_prime_root:.4f}")
print(f"Зменшення нечистоти (ΔG): {delta_G_root:.4f}\n")

n_l_node = tree_model.tree_.n_node_samples[1]
G_l_node = tree_model.tree_.impurity[1]
n_L3 = tree_model.tree_.n_node_samples[3]
G_L3 = tree_model.tree_.impurity[3]
n_R4 = tree_model.tree_.n_node_samples[4]
G_R4 = tree_model.tree_.impurity[4]

G_prime_left = (n_L3 / n_l_node) * G_L3 + (n_R4 / n_l_node) * G_R4
delta_G_left = G_l_node - G_prime_left

print("--- ЛІВА ДИТИНА КОРЕНЯ ---")
print(f"Індекс Джині кореня (G): {G_l_node:.4f}")
print(f"Зважений Джині (G'): {G_prime_left:.4f}")
print(f"Зменшення нечистоти (ΔG): {delta_G_left:.4f}\n")

n_r_node = tree_model.tree_.n_node_samples[2]
G_r_node = tree_model.tree_.impurity[2]
n_L5 = tree_model.tree_.n_node_samples[5]
G_L5 = tree_model.tree_.impurity[5]
n_R6 = tree_model.tree_.n_node_samples[6]
G_R6 = tree_model.tree_.impurity[6]

G_prime_right = (n_L5 / n_r_node) * G_L5 + (n_R6 / n_r_node) * G_R6
delta_G_right = G_r_node - G_prime_right

print("--- ПРАВА ДИТИНА КОРЕНЯ ---")
print(f"Індекс Джині кореня (G): {G_r_node:.4f}")
print(f"Зважений Джині (G'): {G_prime_right:.4f}")
print(f"Зменшення нечистоти (ΔG): {delta_G_right:.4f}")
```
